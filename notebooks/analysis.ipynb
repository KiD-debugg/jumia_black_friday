##### Pre-Cleaning EDA/Basic checks

```python
def basic_checks(df):
    print("-"* 40, "SHAPE", "-"*40)
    print(df.shape)
    
    print("-"* 40, "DUPLICATES", "-"*40)
    print(df.duplicated().sum())
    
    print(("-"* 40, "COLUMNS", "-"*40))
    print(df.columns.to_list())
    
    print(("-"* 40, "MISSING VALUES", "-"*40))
    print(df.isna().sum()) #The isnull() function is simply an alias for isna(). They point to the exact same underlying code, perform the same operation, and deliver identical performance. Both methods are used to detect missing, null, or NaN values in DataFrames and Series, returning a boolean mask of the same shape
    
    print(("-"* 40, "BASIC INFO", "-"*40))
    print(df.info())
```

```python
basic_checks(jumia_data)
```

#### Data Visualization with seaborn and matplotlib
##### 1. Bar Chart 
Example analysis: How many products are in each category?

```python
#Matplotlib: 
#1st step is to count
counts = jumia_data["category"].value_counts()
counts

#2. Create a bar plot
plt.figure(figsize=(10,5)) #This is optional to have because matplotlib has its own default values ie (6.4in by 8.5 in)
plt.bar(counts.index, counts.values)
plt.title("Products Per Category")
plt.xlabel("Category")
plt.ylabel("Number Of Products")
plt.xticks(rotation=45)
plt.show()
```

```python
#Using Seaborn
plt.figure(figsize=(10,5))
sns.countplot(data=jumia_data, x="category", order = counts.index)
plt.title("Products Per Category")
plt.xlabel("Category")
plt.ylabel("Number Of Products")
plt.xticks(rotation=45)
plt.show()
```

##### Line Chart
Analysis Question:. Does jumia show biggest discounts at the top of the black friday page

```python
main_black_friday = jumia_data[jumia_data["listing_source"] == "Main Black Friday page"] # Products from main page

# Calculate average discount for each page
avg_discount = main_black_friday.groupby("listing_page")["discount_pct"].mean()

avg_discount
```

```python
# Matplotlib
plt.figure(figsize=(12, 8))
plt.plot(avg_discount.index, avg_discount.values, marker="o")
plt.title("Average discount per page")
plt.xlabel("Listing page")
plt.ylabel("Average Discount")
plt.xticks(range(1, 55, 1), rotation=90)
plt.yticks(range(25, 60, 1))
plt.show()
```

```python
# Seaborn
plt.figure(figsize=(12, 8))
sns.lineplot(data=main_black_friday, x="listing_page", y="discount_pct", marker="s", errorbar=None) # estimator="median"
plt.title("Average discount per page")
plt.xlabel("Listing page")
plt.ylabel("Average Discount")
plt.xticks(range(1, 55, 1), rotation=90)
plt.yticks(range(25, 60, 1))
plt.show()
```

```python
avg_current_price = main_black_friday.groupby("listing_page")["current_price_ksh"].mean()
# Matplotlib
plt.figure(figsize=(20, 6))
plt.plot(avg_current_price.index, avg_current_price.values, marker="o")
plt.title("Avg current price per listing page")
plt.xlabel("Listing Page")
plt.xticks(range(1, 55, 1))
plt.ylabel("Average current price")
plt.show()
```

```python
# Seaborn
plt.figure(figsize=(20, 6))
sns.lineplot(data=main_black_friday, x="listing_page", y="current_price_ksh", marker="s", errorbar=None)
plt.title("Avg current price per listing page")
plt.xlabel("Listing Page")
plt.xticks(range(1, 55, 1))
plt.ylabel("Average current price")
plt.show()
```

#### Pie Chart/Donut Chart
What share of products come from official stores?

```python
# Matplotlib
# Pie Chart

share = jumia_data["is_official_store"].value_counts()
share

plt.pie(share, labels=["Non-official store", "Official store"], autopct="%1.0f%%", explode=(0, 0.1))
plt.title("Stores Distribution")
plt.show()
```

```python
# Donut Chart
plt.pie(share, labels=["Non-official store", "Official store"], autopct="%1.0f%%", 
        explode=(0, 0.1), wedgeprops={"width": 0.5})
plt.title("Stores Distribution")
plt.show()
```
